```python
'''
Practise Pandas/Plotting
'''
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
#Number 1

sales_df = pd.read_csv('sales.csv')
for i in range(1,len(sales_df.columns)):
    mean = np.mean(sales_df.iloc[:,i])
    print(mean)
for i in range(1,len(sales_df.columns)):
    xpoints = sales_df.iloc[:,0]
    ypoints = sales_df.iloc[:,i]
    plt.plot(xpoints,ypoints, label = f"{sales_df.columns[i]}")
plt.legend(fontsize = 8)
plt.title("hello")
plt.xlabel('hello')
```
