```python
#sort a list of lists 
# - given list of lists(assume integers), sort each sublist in decreasing order
# - cannot use any built-in functions
# - must define a sort that allows to sort both in increasing or decreasing
# - what is the time and space complexity?
# HOMEWORK

lst = [[1,5,7],[],[1,6,5,77]]
lst2 = [1,5,3,4]

def partition(array, low, high,mode):
    pivot = array[high]

    # pointer for greater element
    i = low - 1
    for j in range(low, high):
        if mode == "ascending":
            if array[j] <= pivot:
                i = i + 1
                (array[i], array[j]) = (array[j], array[i])
                
        else:
            if array[j] >= pivot:
                i = i + 1
                (array[i], array[j]) = (array[j], array[i])
    (array[i + 1], array[high]) = (array[high], array[i + 1])
    return i + 1
def quickSort(array, low, high, mode):
    if low < high:
        pi = partition(array, low, high, mode)
        quickSort(array, low, pi - 1, mode)
        quickSort(array, pi + 1, high, mode)

#quickSort(lst2, 0, len(lst2) - 1)
#print(lst2)

def mySort(lst, mode):
    for row in lst:
        if len(row) > 1:
            quickSort(row, 0, len(row)-1, mode)

mySort(lst, "descending")
print(lst)




    


# given an n by m matrix, write code to transpose it
# cannot use any built-in functions
# columns change into rows and rows change into columns
#Try to build it with list comprehension







# HOMEWORK      

    
    

```
