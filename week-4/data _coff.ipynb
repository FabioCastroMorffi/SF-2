```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
coffee_df = pd.read_csv('data_for_coffee.csv')
coffee_df[["Time (s) Run #2","Position (m) Run #2"]].plot(x = "Time (s) Run #2" , y = "Position (m) Run #2", kind="line", grid=True)

```

```python

def minVal(coffee_df):
    position_df = coffee_df['Position (m) Run #2']
    min = int(position_df.at[0])

    for i in range(1, len(position_df)):
        curr = int(position_df.at[i])
        if curr < min:
            min = curr
    return min
answer = minVal(coffee_df)
print(answer)
```
